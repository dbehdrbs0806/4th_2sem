```python
import matplotlib.pyplot as plt
plt.rc('font', family='Malgun Gothic')  
plt.rcParams['axes.unicode_minus'] = False 


iris['sepal_length'][:10].plot(kind='bar', rot=0)
plt.title('Iris Sepal Length')
plt.xlabel('Data')
plt.ylabel('꽃받침의 길이')
plt.show()
```
